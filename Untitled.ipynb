```python
import json, time
from pathlib import Path
import cv2, numpy as np, matplotlib.pyplot as plt
from rtmlib import Wholebody, draw_skeleton
```

```python
video_path = "/home/topiary-pc/Projects/MovesMethod/Example1.mov"
```

```python
video_loader = cv2.VideoCapture(video_path)
```

```python
model = Wholebody(mode="lightweight", backend="onnxruntime", device="cpu")
```

```python
n_frame = 0
plt.figure(figsize=(6, 10))
while True:
    ok, img = video_loader.read()
    if not ok: break

    dets = model.det_model(img)
    box = max(dets, key=lambda b: (b[2]-b[0])*(b[3]-b[1]))
    
    k, s = model.pose_model(img, bboxes=[box])
    vis = draw_skeleton(img.copy(), k, s, kpt_thr=0.3)
    print(dets, k.shape, s.shape)
    
    
    n_frame += 1

    if n_frame == 200:
        plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))   
        plt.axis("off")
        break
plt.show()
```

```python
# foot keypoints: ankle / heel / big toe / small toe, both sides
FOOT = {15: ("L ankle",     (0, 0, 255)),   17: ("L big toe",  (0, 255, 0)),
        18: ("L small toe", (0, 255, 255)), 19: ("L heel",     (255, 0, 255)),
        16: ("R ankle",     (0, 0, 255)),   20: ("R big toe",  (0, 255, 0)),
        21: ("R small toe", (0, 255, 255)), 22: ("R heel",     (255, 0, 255))}
FRAME_TO_CHECK = 900          # ~30 s: side-view squat hold

video_loader.set(cv2.CAP_PROP_POS_FRAMES, FRAME_TO_CHECK)
ok, img = video_loader.read()
dets = model.det_model(img)
box = max(dets, key=lambda b: (b[2]-b[0])*(b[3]-b[1]))
keypoints, scores = model.pose_model(img, bboxes=[box])
k, s = keypoints[0], scores[0]

vis = draw_skeleton(img.copy(), keypoints, scores, kpt_thr=0.3)
for j, (label, colour) in FOOT.items():
    p = tuple(map(int, k[j]))
    left = j in (15, 17, 18, 19)                 # left-side points: number above, right-side: below
    cv2.circle(vis, p, 10, colour, -1, cv2.LINE_AA)
    cv2.circle(vis, p, 10, (255, 255, 255), 2, cv2.LINE_AA)
    cv2.putText(vis, str(j), (p[0] - 10, p[1] - 16 if left else p[1] + 30),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, colour, 2, cv2.LINE_AA)

# zoom on the feet
ys = [int(k[j][1]) for j in FOOT]; xs = [int(k[j][0]) for j in FOOT]
crop = vis[max(min(ys)-150, 0):min(max(ys)+60, vis.shape[0]), max(min(xs)-100, 0):min(max(xs)+120, vis.shape[1])]

fig, ax = plt.subplots(1, 3, figsize=(18, 9), gridspec_kw={"width_ratios": [1, 1.6, 0.8]})
ax[0].imshow(cv2.cvtColor(vis,  cv2.COLOR_BGR2RGB)); ax[0].axis("off")
ax[1].imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB)); ax[1].axis("off")
legend = "\n".join(f"{j:>2}  {label:<12} score {s[j]:.2f}" for j, (label, _) in sorted(FOOT.items()))
ax[2].text(0, 0.5, legend, family="monospace", fontsize=13, va="center"); ax[2].axis("off")
plt.tight_layout(); plt.show()
```
