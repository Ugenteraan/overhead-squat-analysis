```python
import json, time
from pathlib import Path
import cv2, numpy as np, matplotlib.pyplot as plt
from rtmlib import Wholebody, draw_skeleton
```

```python
video_path = "/home/topiary-pc/Projects/MovesMethod/Example1.mov"
```

```python
video_loader = cv2.VideoCapture(video_path)
```

```python
model = Wholebody(mode="lightweight", backend="onnxruntime", device="cpu")
```

```python
n_frame = 0
plt.figure(figsize=(6, 10))
while True:
    ok, img = video_loader.read()
    if not ok: break

    dets = model.det_model(img)
    box = max(dets, key=lambda b: (b[2]-b[0])*(b[3]-b[1]))
    
    k, s = model.pose_model(img, bboxes=[box])
    vis = draw_skeleton(img.copy(), k, s, kpt_thr=0.3)
    print(dets, k.shape, s.shape)
    
    
    n_frame += 1

    if n_frame == 200:
        plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))   
        plt.axis("off")
        break
plt.show()
```

```python
HIGHLIGHT = {5: ("5 = left_shoulder",  (0, 0, 255)),     # red   (OpenCV colours are BGR)
               6: ("6 = right_shoulder", (255, 0, 0))}     # blue
FRAME_TO_CHECK = 420          # ~14 s: frontal squat hold, both shoulders clearly visible

video_loader.set(cv2.CAP_PROP_POS_FRAMES, FRAME_TO_CHECK)
ok, img = video_loader.read()

if ok:
  dets = model.det_model(img)
  if len(dets):
      box = max(dets, key=lambda b: (b[2]-b[0])*(b[3]-b[1]))
      
      keypoints, scores = model.pose_model(img, bboxes=[box])
      k, s = keypoints[0], scores[0]                   # drop the person dimension -> (133, 2), (133,)
      
      vis = img.copy()
      
      # all body + feet points, small and grey, for context
      for j in range(0, 23):
          if s[j] > 0.3:
              cv2.circle(vis, tuple(map(int, k[j])), 4, (180, 180, 180), -1, cv2.LINE_AA)

      # the points under test
      for j, (label, colour) in HIGHLIGHT.items():
          p = tuple(map(int, k[j]))
          cv2.circle(vis, p, 12, colour, -1, cv2.LINE_AA)
          cv2.circle(vis, p, 12, (255, 255, 255), 2, cv2.LINE_AA)      # white ring so it shows on any background
          cv2.putText(vis, f"{label}  ({s[j]:.2f})", (p[0] + 16, p[1] + 6),
                      cv2.FONT_HERSHEY_SIMPLEX, 0.7, colour, 2, cv2.LINE_AA)
      
      # show once, after everything is drawn
      plt.figure(figsize=(8, 14))
      plt.imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB))
      plt.axis("off")
      plt.show()
  else:
      print("no person detected in this frame")
else:
  print("could not read frame", FRAME_TO_CHECK)
```
